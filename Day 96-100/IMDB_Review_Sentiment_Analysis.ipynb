Preprocessing


*   lower casing
*   remove html tags
*   remove punctuation
*   Stop words
*   Stemming/lematization



```python
# To lowercase
df['review']= df['review'].str.lower()
df
```

```python
# removing html tags
import re
def remove_html(text):
    pattern = re.compile('<.*?>')
    return pattern.sub('', text)
```

```python
df['review']=df['review'].apply(remove_html)
df
```

```python
#Removing punctuation
import string
def remove_punc(text):
  return text.translate(str.maketrans('','', string.punctuation))
```

```python
df['review']=df['review'].apply(remove_punc)
df
```

```python
#removing stop words
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))

    words = text.split()

    filtered_words = [
        word for word in words
        if word.lower() not in stop_words
    ]

    return " ".join(filtered_words)
```

```python
df['review']=df['review'].apply(remove_stopwords)
df
```

```python
import nltk
nltk.download('wordnet')
```

```python
from nltk.stem import WordNetLemmatizer
wl=WordNetLemmatizer()

def lem(text):
  words=text.split()
  words=([wl.lemmatize(word)for word in words])
  return " ".join(words)
```

```python
df['review']=df['review'].apply(lem)
df['review'][0]
```

```python
x=df.drop('sentiment', axis=1)
y=df['sentiment']
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()

y=le.fit_transform(y)
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y, test_size=0.2, random_state=51)
```

```python
len(y_test)
```

```python
#Using BOW or TFIDF

from sklearn.feature_extraction.text import CountVectorizer

cv=CountVectorizer()#in this we can give max_fetures=3000 like this, it will take most repeatable 3000 words/features to train the model
# we can alsso use ngrams but space will be exhausted because of exponential increase in features
# we can also use TF-IDF instead of Bow
x_train_bow=cv.fit_transform(x_train['review']).toarray()
x_test_bow=cv.transform(x_test['review']).toarray()

```

```python
from sklearn.naive_bayes import GaussianNB

model=GaussianNB()
```

```python
model.fit(x_train_bow, y_train)
```

```python
ypred=model.predict(x_test_bow)
```

```python
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report
```

```python
print(accuracy_score(ypred, y_test))
```

```python
print(confusion_matrix(ypred, y_test))
```

```python
print(classification_report(ypred, y_test))
```

```python
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier()

rf.fit(x_train_bow, y_train)
```

```python
y_pred=rf.predict(x_test_bow)

print(accuracy_score(y_pred, y_test)*100)
```

```python
import gensim
import numpy as np
import nltk
from nltk import sent_tokenize
from gensim.utils import simple_preprocess
nltk.download('punkt_tab')
```

```python
data=[]

for word in df['review']:
  rs=sent_tokenize(word)
  for sent in rs:
    data.append(simple_preprocess(sent))
```

```python
len(data)
```

```python
model=gensim.models.Word2Vec(
    window=5,
    min_count=2
)
```

```python
model.build_vocab(data)
```

```python
model.train(data, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
def doc_vector(doc):
  doc=[word for word in doc.split() if word in model.wv.index_to_key]
  return np.mean(model.wv[doc], axis=0)
```
