```python
import pandas as pd
```

```python
df=pd.read_csv('/content/7817_1.csv')
df
```

```python
df['reviews.rating'].value_counts()
```

```python
def sentiment(text):
  if text<3:
    return "Negative"
  elif text==3:
    return "Neutral"
  else:
    return "Positive"
```

```python
df['reviews.rating']=df['reviews.rating'].apply(sentiment)
df
```

```python
df2=df[['reviews.text', 'reviews.rating']].copy()
```

```python
df2
```

```python
df2['reviews.text']=df2['reviews.text'].str.lower()
```

```python
df2
```

```python
import string
def remove_punc(text):
  return text.translate(str.maketrans('','', string.punctuation))
```

```python
df2['reviews.text']=df2['reviews.text'].apply(remove_punc)
df2
```

```python
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))

    words = text.split()

    filtered_words = [
        word for word in words
        if word.lower() not in stop_words
    ]

    return " ".join(filtered_words)
```

```python
df2['reviews.text']=df2['reviews.text'].apply(remove_stopwords)
df2
```

```python
import nltk
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
wl=WordNetLemmatizer()
def lemmatize(text):
  text=text.split()
  text=[wl.lemmatize(word, pos='v') for word in text]
  return ' '.join(text)
```

```python
df2['reviews.text']=df2['reviews.text'].apply(lemmatize)
df2
```

```python
x=df2['reviews.text']
y=df2['reviews.rating']
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
y=le.fit_transform(y)
y
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test=train_test_split(x,y, test_size=0.25, random_state=5)
```

# **First Approach: TFIDF + Logistic Regression**

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tf=TfidfVectorizer()

xtrain=tf.fit_transform(x_train).toarray()
xtest=tf.transform(x_test).toarray()
```

```python
xtrain
```

```python
from sklearn.linear_model import LogisticRegression

lr=LogisticRegression()
```

```python
lr.fit(xtrain, y_train)
```

```python
ypred=lr.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(accuracy_score(ypred,y_test)*100)
```

```python
import gensim
from gensim.utils import simple_preprocess
from nltk import sent_tokenize
nltk.download('punkt_tab')
```

```python
data=[]

for word in df2['reviews.text']:
  rs=sent_tokenize(word)
  for s in rs:
    data.append(simple_preprocess(s))
```

```python
data
```

```python
model=gensim.models.Word2Vec(
    window=5
)
```

```python
model.build_vocab(data)
```

```python
model.train(data, total_examples=model.corpus_count, epochs=model.epochs)
```

```python
import numpy as np
def doc_vector(doc):
  doc=[word for word in doc.split() if word in model.wv.index_to_key]
  if len(doc) == 0:
        return np.zeros(model.vector_size)

  return np.mean(model.wv[doc], axis=0)
```

```python
x1=[]
for doc in df2['reviews.text'].values:
  x1.append(doc_vector(doc))
```

```python
#sum(df2['reviews.text'].str.strip() == "")
```

```python
x1[0]
```

```python
y1=y.copy()
```

```python
y1
```

```python
xtr, xtt, ytr, ytt=train_test_split(x1,y1, test_size=0.2, random_state=5)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf=RandomForestClassifier()
```

```python
rf.fit(xtr,ytr)
```

```python
yp=rf.predict(xtt)
```

```python
print(accuracy_score(yp, ytt))
```
