```python
import pandas as pd
```

```python
df=pd.read_csv('/content/spam.csv', encoding='latin-1')
df
```

```python
df.drop(df[['Unnamed: 2',	'Unnamed: 3',	'Unnamed: 4']], axis=1, inplace=True)
```

```python
df
```

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
from sklearn.preprocessing import LabelEncoder

le=LabelEncoder()
```

```python
df['v1']=le.fit_transform(df['v1'])
df
```

```python
df['v2']=df['v2'].str.lower()
df
```

```python
import string

def remove_punc(text):
  return text.translate(str.maketrans(' ', ' ', string.punctuation))
```

```python
df['v2']=df['v2'].apply(remove_punc)
df
```

```python
#removing stop words
import nltk
from nltk.corpus import stopwords

nltk.download('stopwords')

def remove_stopwords(text):
    stop_words = set(stopwords.words('english'))

    words = text.split()

    filtered_words = [
        word for word in words
        if word.lower() not in stop_words
    ]

    return " ".join(filtered_words)
```

```python
df['v2']=df['v2'].apply(remove_stopwords)
df
```

```python
import nltk
from nltk import WordNetLemmatizer
nltk.download('wordnet')
```

```python
wl=WordNetLemmatizer()
def lemmatize(text):
  text=text.split()
  text=[wl.lemmatize(word, pos='v') for word in text]
  return ' '.join(text)
```

```python
df['v2']=df['v2'].apply(lemmatize)
```

```python
x=df['v2']
y=df['v1']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB, MultinomialNB, BernoulliNB
```

```python
xtr, xtt, ytr, ytt=train_test_split(x,y, test_size=0.2, random_state=5)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tf=TfidfVectorizer()

xtrain=tf.fit_transform(xtr).toarray()
xtest=tf.transform(xtt).toarray()
```

```python
gb=GaussianNB()
mb=MultinomialNB()
bb=BernoulliNB()
```

```python
gb.fit(xtrain,ytr)
y1=gb.predict(xtest)
```

```python
mb.fit(xtrain, ytr)
y2=mb.predict(xtest)
```

```python
bb.fit(xtrain, ytr)
y3=bb.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
```

```python
print(f"GaussianNB: {accuracy_score(y1,ytt)*100:.2f}%")
print(f"MultinomialNB: {accuracy_score(y2,ytt)*100:.2f}%")
print(f"BernouliNB: {accuracy_score(y3,ytt)*100:.2f}%")
```
