```python
import tensorflow
from tensorflow import keras
from keras.models import Model
```

```python
model=Model(inputs=x, outputs=[op1, op2])
```

```python
from keras.layers import *
```

```python
x=Input(shape=(3,))
```

```python
h1= Dense(128, activation='relu')(x)
h2= Dense(64, activation='relu')(h1)

op1= Dense(1, activation='linear')(h2)
op2= Dense(1, activation='sigmoid')(h2)
```

```python
model.summary()
```

```python
from keras.utils import plot_model
plot_model(model, show_shapes=True)
```

```python
ip1=Input(shape=(128,))
ip2=Input(shape=(32,))
```

```python
# for first branch

x1= Dense(64, activation='relu')(ip1)
x2= Dense(32, activation='relu')(x1)
x3= Dense(4, activation='relu')(x2)
```

```python
#for Second branch

y1= Dense(8, activation='relu')(ip2)
y2= Dense(4, activation='relu')(y1)
```

```python
combine= concatenate([x3,y2])
```

```python
z1= Dense(2, activation='relu')(combine)
z2= Dense(1, activation='sigmoid')(z1)
```

```python
model1=Model(inputs=[ip1, ip2], outputs=z2)
```

```python
plot_model(model1)
```
