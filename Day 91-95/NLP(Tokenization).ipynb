# **Tokenization**

```python
import nltk
nltk.download('punkt')
nltk.download('punkt_tab')
```

```python
# this is for sentence tokenization from paragraph
from nltk.tokenize import sent_tokenize
```

```python
corpus = """
India is a diverse country with a rich cultural heritage.
It is known for its festivals, traditions, and historical monuments.
 Millions of tourists visit India every year to explore its beauty.
 The country has made significant progress in science and technology.
 India is also one of the fastest-growing economies in the world.
"""
```

```python
sent_tokenize(corpus)
# using . and ! make the sentences alag and seperated
```

```python
# For word tokenization from Paragraph
```

```python
from nltk.tokenize import word_tokenize
```

```python
word_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tk=TreebankWordTokenizer()
```

```python
tk.tokenize(corpus)
```
