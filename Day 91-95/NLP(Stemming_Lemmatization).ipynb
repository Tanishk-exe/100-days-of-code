```python
import nltk
from nltk.stem import PorterStemmer
```

```python
# Porter Stemmer algo
st=PorterStemmer()
```

```python
words = [
    "playing",
    "played",
    "plays",
    "studying",
    "studies",
    "studied",
    "running",
    "runs",
    "ran",
    "eating",
    "eats",
    "ate",
    "driving",
    "drove",
    "driven"
]
```

```python
for word in words:
  print(word +"--->" + st.stem(word))
```

```python
from nltk.stem import LancasterStemmer
```

```python
lt=LancasterStemmer()
```

```python
for word in words:
  print(word +"--->" + lt.stem(word))
```

```python
#RegexpStemmer
from nltk.stem import RegexpStemmer
```

```python
rt=RegexpStemmer('ing$|able$|ies$')
```

```python
rt.stem('playing')
```

```python
# SnowBall Stemmer
from nltk.stem import SnowballStemmer
```

```python
sbt=SnowballStemmer('english')
```

```python
for word in words:
  print(word +"--->" + sbt.stem(word))
```

# **Lemmatization**

```python
nltk.download('wordnet')
from nltk.stem import WordNetLemmatizer
```

```python
wl=WordNetLemmatizer()
```

**pos means it takes what kind of word is there to lemmatize**

noun-n

verb-v

adjective-a

adverb-r


```python
for word in words:
  print(word +"--->" + wl.lemmatize(word, pos='v'))
```
